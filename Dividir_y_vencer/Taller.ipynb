# 1. Ordenar el arreglo

## Descripción del Problema

Dada una lista de números enteros `nums`, ordena el arreglo de forma ascendente utilizando el algoritmo **QuickSort**.

Debes implementar una solución basada en el paradigma **Divide y Vencerás** siguiendo la estrategia de partición por pivote.

### Especificaciones del Algoritmo:
1.  **Estrategia de Pivote:** Para este ejercicio, utiliza siempre el **último elemento** del sub-arreglo actual como pivote.
2.  **Partición:** Debes reordenar el arreglo de tal manera que todos los elementos menores o iguales al pivote queden a su izquierda y los elementos mayores a su derecha.
3.  **Recursión:** Aplica el mismo proceso de forma recursiva para las mitades generadas hasta que el arreglo esté completamente ordenado.

### Ejemplo 1:
**Input:** `nums = [5, 2, 3, 1]`  
**Output:** `[1, 2, 3, 5]`  
**Explicación:** - Pivote inicial: `1`.
- Tras la partición: `izq = []`, `der = [5, 2, 3]`.
- El proceso continúa recursivamente.

### Ejemplo 2:
**Input:** `nums = [5, 1, 1, 2, 0, 0]`  
**Output:** `[0, 0, 1, 1, 2, 5]`

### Restricciones:
* `1 <= nums.length <= 5 * 10^4`
* `-5 * 10^4 <= nums[i] <= 5 * 10^4`
* **Reto:** Intenta realizar la partición *in-place* para optimizar el uso de memoria.

````python
lista = [6,1,2,5,3]

def dividir(lista):
    if len(lista) <= 1:
        print(lista)
        return lista
    
    pivot = lista.pop()
    izq = []
    der = []
    for num in lista:
        if pivot >= num:
            izq.append(num)
            continue
        der.append(num)
    
    i = dividir(izq)
    d = dividir(der)

    lista = i + [pivot] + d

    return lista
    
dividir(lista)




````

# 2. Encuentra la primera y ultima posicion en un arreglo ordenado

## Descripción del Problema

Dado un arreglo de números enteros `nums` ordenado en orden no decreciente (donde cada número representa el ID de un grupo de estudiantes), encuentra los índices de la **primera** y la **última** posición de un grupo dado `target`.

Si el grupo `target` no se encuentra en el arreglo, devuelve `[-1, -1]`.

### Ejemplo 1:
**Input:** `nums = [1, 1, 2, 2, 2, 2, 3], target = 2`  
**Output:** `[2, 5]`  
**Explicación:** El grupo 2 comienza en el índice 2 y termina en el índice 5. El total de estudiantes es $5 - 2 + 1 = 4$.

### Ejemplo 2:
**Input:** `nums = [1, 1, 2, 2, 2, 2, 3], target = 4`  
**Output:** `[-1, -1]`

### Restricciones:
* `0 <= nums.length <= 10^5`
* `-10^9 <= nums[i] <= 10^9`
* `nums` es un arreglo ordenado de forma ascendente.

````python
arr = [4,4,4,4,5,5,5,5,5]

def encontrar_ultimo(arr, k, i = 0, j = None):
    if j is None:
        j = len(arr) - 1
    mid = (j + i)// 2

    if i == j:
        if k != arr[i]:
            return -1
        return i
    
    if k < arr[mid]:
        return encontrar_ultimo(arr, k, i, mid - 1)
    if k >= arr[mid]:
        return encontrar_ultimo(arr, k, mid + 1 , j)
    
    
def encontrar_primero(arr, k, i = 0, j = None):
    if j is None:
        j = len(arr) - 1


    print(1)
    if i == j:
        if k != arr[i]:
            return -1
        return i
    
    mid = (j + i) // 2
    if k <= arr[mid]:
        return encontrar_primero(arr, k, i, mid)
    if k > arr[mid]:
        return encontrar_primero(arr, k, mid + 1, j)
    
def sacar_indices(arr, k):
    izquierda = encontrar_primero(arr, k)
    derecha = encontrar_ultimo(arr, k)
    return izquierda, derecha

sacar_indices(arr, 5)
````

# 3. Verificación de Tamaño de Grupo

## Descripción del Problema

El departamento académico ha establecido una regla estricta: para el próximo proyecto integrador, un grupo solo es válido si tiene **exactamente $K$ integrantes**.

Utilizando la lógica que desarrollaste en el **Ejercicio 2** (Búsqueda Binaria para encontrar los límites de un grupo), crea una función que determine si un grupo específico cumple con esta condición.

### Tarea:
Implementa una función que reciba el arreglo ordenado de estudiantes, el ID del grupo a verificar y el número esperado de integrantes ($K$). La función debe devolver `True` si el grupo tiene exactamente esa cantidad de estudiantes y `False` en caso contrario.

### Ejemplo 1:
**Input:** `nums = [1, 1, 2, 2, 2, 2, 3], target_group = 2, k = 4`  
**Output:** `True`  
**Explicación:** El grupo 2 aparece 4 veces. Como $k = 4$, el resultado es verdadero.

### Ejemplo 2:
**Input:** `nums = [1, 1, 2, 2, 2, 2, 3], target_group = 1, k = 4`  
**Output:** `False`  
**Explicación:** El grupo 1 solo tiene 2 integrantes.

````python
def buscar_menor(arr, target_group, i = 0, j = None):
    if j is None:
        j = len(arr) - 1

    if i == j:
        if arr[i] != target_group:
            return False
        return i
    mid = (j + i) // 2
    
    if target_group <= arr[mid]:
        return buscar_menor(arr, target_group, i, mid )
    if target_group > arr[mid]:
        return buscar_menor(arr, target_group, mid + 1, j)
    

def buscar_mayor(arr, target_group, i = 0, j = None):
    if j is None:
        j = len(arr) - 1

    mid = (j + i + 1) // 2
    if i == j:
        if target_group != arr[i]:
            return False
        return i
    
    if target_group < arr[mid]:
        return buscar_mayor(arr, target_group, i, mid - 1)
    else:
        return buscar_mayor(arr, target_group, mid , j)


def verificar_grupo(nums, target_group, k):
    izquierda = buscar_menor(nums, target_group)
    derecha = buscar_mayor(nums, target_group)

    if izquierda is False or derecha is False: 
        return False

    return (derecha - izquierda + 1) == k


nums = [2, 2, 2, 5]
print(verificar_grupo(nums, 2, 4))  # → True
print(verificar_grupo(nums, 1, 4))  # → False
print(verificar_grupo(nums, 5, 1))  # → False
````
