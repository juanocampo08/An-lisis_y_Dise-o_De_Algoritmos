## Ejercicio 1 — Fuerza Bruta: Contraseña Corporativa

La empresa **SeguraTech** está migrando su sistema de autenticación y necesita validar contraseñas generadas automáticamente. El sistema recibe un conjunto de caracteres únicos y debe encontrar todas las permutaciones de longitud exacta `k` que cumplan **simultáneamente** con la política de seguridad corporativa:

- Contiene al menos una letra mayúscula.
- Contiene al menos un dígito numérico.
- Contiene al menos un símbolo especial (cualquier carácter que no sea letra ni dígito).

Dado un conjunto de caracteres y un entero `k`, retorna la lista de todas las contraseñas válidas en cualquier orden.

**Ejemplo:**

```
caracteres = ['A', '3', '@', 'b']
k = 3
```

Algunas permutaciones posibles son `['A', '3', '@']`, `['3', 'b', 'A']`, `['@', 'A', '3']`, entre muchas otras. De todas ellas, solo deben retornarse las que cumplan las tres reglas al mismo tiempo. Por ejemplo, `['A', '3', '@']` **sí es válida** porque tiene mayúscula (`A`), dígito (`3`) y símbolo (`@`). En cambio, `['A', '3', 'b']` **no es válida** porque no tiene símbolo especial.

## Ejercicio 2 — Greedy: Cobertura de Noticias

El portal **ElTiempo Digital** tiene un equipo editorial con capacidad limitada: solo puede publicar exactamente `k` notas periodísticas por semana. Cada mañana, los reporteros entregan un pool de notas nuevas, cada una con un puntaje de impacto estimado. El editor debe seleccionar, día a día, la nota de **mayor impacto disponible** hasta completar los `k` slots de la semana.

Se te entrega una lista de listas, donde cada sublista representa las notas que llegan en ese día con sus respectivos impactos. Retorna la lista con los impactos de las `k` notas publicadas, en el orden en que fueron seleccionadas. Si en algún momento no hay notas disponibles, el editor espera al siguiente día. Si al final de la semana hay menos de `k` notas disponibles en total, publica todas las que haya.

**Ejemplo:**

```
notas_por_dia = [[5, 3], [9, 1], [7]]
k = 4
```

El lunes llegan notas de impacto `5` y `3`. El mejor disponible es `9`... pero aún no ha llegado, así que se publica el `5`. El martes llegan `9` y `1`; el pool disponible ahora tiene `[3, 9, 1]`, se publica el `9`. El miércoles llega el `7`; el pool tiene `[3, 1, 7]`, se publica el `7`. Queda un slot: se publica el `3`.

```
Resultado: [5, 9, 7, 3]
```

````python
import heapq

def mayor_impacto(notas: list[list[int]], k: int):
    # max-heap simulado con negativos
    pool = []
    heapq.heapify(pool)

    publicadas = []

    # Día a día: entra lo nuevo, sale la mejor disponible
    for notas_del_dia in notas:
        for impacto in notas_del_dia:
            heapq.heappush(pool, -impacto)   # push

        if len(publicadas) < k and pool:
            mejor = -heapq.heappop(pool)     # pop
            publicadas.append(mejor)

    # Si faltan slots, seguir publicando del pool acumulado
    while len(publicadas) < k and pool:
        publicadas.append(-heapq.heappop(pool))  # pop

    return publicadas


notas_por_dia = [[5, 3], [9, 1],[7]]
k = 4

print(mayor_impacto(notas_por_dia, k))
````

````python
def puedo_colocar(asignacion, capacidades, deps_por_col, grupo, r, c):
    nombre, depto, tam = grupo

    if asignacion[r][c] is not None:
        return False

    if tam > capacidades[r][c]:
        return False

    if depto in deps_por_col[c]:
        return False

    return True


def backtrack_asignar(i, grupos, asignacion, capacidades, deps_por_col, n, m):
    if i == len(grupos):
        return True

    grupo = grupos[i]
    nombre, depto, tam = grupo

    r = 0
    while r < n:
        c = 0
        while c < m:
            if puedo_colocar(asignacion, capacidades, deps_por_col, grupo, r, c):
                # elegir
                asignacion[r][c] = nombre
                deps_por_col[c].add(depto)

                # explorar
                if backtrack_asignar(i + 1, grupos, asignacion, capacidades, deps_por_col, n, m):
                    return True

                # deshacer
                asignacion[r][c] = None
                deps_por_col[c].remove(depto)
            c += 1
        r += 1

    return False


def asignar_aulas(capacidades, grupos):
    n = len(capacidades)
    if n == 0:
        return []

    m = len(capacidades[0])

    # crear matriz asignacion sin list comprehension
    asignacion = []
    i = 0
    while i < n:
        fila = []
        j = 0
        while j < m:
            fila.append(None)
            j += 1
        asignacion.append(fila)
        i += 1

    # crear deps_por_col sin list comprehension
    deps_por_col = []
    j = 0
    while j < m:
        deps_por_col.append(set())
        j += 1

    if backtrack_asignar(0, grupos, asignacion, capacidades, deps_por_col, n, m):
        return asignacion
    return None


# ejemplo
capacidades = [[30, 25],
               [20, 35]]

grupos = [
    ("G1", "Sistemas", 20),
    ("G2", "Sistemas", 25),
    ("G3", "Industrial", 30),
]

print(asignar_aulas(capacidades, grupos))
````
